```python
import sympy as sp
print("sympy version:", sp.__version__)
```

Define variables 

```python
x,y = sp.symbols('x y')
```

```python
x+2*y
```

linear equation

```python
eq = sp.Eq(2*x+5,15)
solution= sp.solve(eq,x)
print("Solution for x:", solution)
```

quadratic equation

```python
eq = sp.Eq(x**2 + 3*x - 4, 0)
solution= sp.solve(eq,x)
solution
```

```python
eq1 = sp.Eq(2*x + y, 10)
eq2 = sp.Eq(x - y, 2)
solution = sp.solve((eq1, eq2), (x, y))
solution
```

Calculus derivative 

```python
f = x**3 + 2*x**2 - x + 5
derivative = sp.diff(f,x)
derivative
```

Integration

```python
sp.integrate(f,x)
```

```python
sp.integrate(x**2, (x, 0, 2))
```

verification System 

```python
def verification(eq , solution):
    lhs = eq.lhs.subs(x , solution)
    rhs = eq.rhs

    return lhs == rhs
```

```python
eq = sp.Eq(2*x+5,15)
verification(eq , 5)
```

Solver function

```python
def solve_equation(equation):

    x = sp.symbols('x')
    lhs , rhs = equation.split('=')
    eq = sp.Eq(sp.simplify(lhs), sp.simplify(rhs))
    solution = sp.solve(eq, x)
    return solution
```

```python
solve_equation("2*x + 5 =15")
```
